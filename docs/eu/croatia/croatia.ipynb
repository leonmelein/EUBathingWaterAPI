```python
import requests
import pandas as pd
```

```python
url = "https://vrtlac.izor.hr/ords/kakvoca/kakvoce_sve_json?p_jezik=en"
```

```python
data = requests.get(url)
```

```python
data = pd.json_normalize(data.json()['markers'])
data = data[[
    'lat', 'lng', 'lsta', 'lpla'
]]
data
```

```python
data.rename(mapper={
    "lsta": "id",
    "lpla": "name",
    "lng": "lon"
}, axis=1, inplace=True)
```

```python
data = data[[
    "id", "name", "lat", "lon"
]]
data
```
