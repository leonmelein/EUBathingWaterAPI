```python
import geopandas as gpd
from requests import Request
from owslib.wfs import WebFeatureService
```

```python
url = "https://www.inspire-geoportal.lt/geoserver/ows"
wfs = WebFeatureService(url=url)
```

```python
def load_wfs_layer(layer_name):
    params = dict(service='WFS', version="2.0.0", request='GetFeature',
      typeName=layer_name, outputFormat='json', srsName="EPSG:4326")
    wfs_request_url = Request('GET', url, params=params).prepare().url
    return gpd.read_file(wfs_request_url) # type: ignore
```

```python
data = load_wfs_layer('pds_22_01:AM.BathingWaters')
data['name'] = data['name'].apply(lambda x: x.title())
data['lat'], data['lon'] = [item.y for item in data['geometry']], [item.x for item in data['geometry']]
data = data[['localid', 'name', 'lat', 'lon']]
data
```
