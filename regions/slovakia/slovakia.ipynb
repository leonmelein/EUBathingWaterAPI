```python
from requests import Request
import geopandas as gpd
```

```python
def loadWFSLayer(layer_name, wfs_url, method='GET'):
    params = dict(service='WFS', version="2.0.0", request='GetFeature',
                        typeName=layer_name, outputFormat='json', srsName="EPSG:4326")
    
    wfs_request_url = Request(method, wfs_url, params=params).prepare().url
    return gpd.read_file(wfs_request_url) # type: ignore
```

```python
url = 'https://geoserver.isuvz.uvzsr.sk/geoserver/wfs'
```

```python
data = loadWFSLayer('hzp:gis_v_r_mapa_ek_sk', url)
```

```python
data['lat'], data['lon'] = data['geometry'].y, data['geometry'].x
data.rename(mapper={
    'Názov': 'name'
}, axis=1, inplace=True)
data['alternate_name'] = data['name']
data = data[[
    'id', 'name', 'alternate_name', 'lat', 'lon'
]]
data
```
